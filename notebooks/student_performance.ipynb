```python
features = [
    'studytime',
    'failures',
    'absences',
    'traveltime',
    'activities',
    'health'
]

X = df[features]
y = df['G3']

print("Input features:")
print(X.head())

print("\nTarget:")
print(y.head())
```

```python
print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training data:")
print("X_train:", X_train.shape)
print("y_train:", y_train.shape)

print("\nTesting data:")
print("X_test:", X_test.shape)
print("y_test:", y_test.shape)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

categorical_features = ['activities']

preprocessor = ColumnTransformer(
    transformers=[
        ('cat', OneHotEncoder(drop='if_binary', handle_unknown='ignore'),
         categorical_features)
    ],
    remainder='passthrough'
)

print("Preprocessing pipeline created successfully.")
```

```python
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression

linear_model = Pipeline(
    steps=[
        ('preprocessor', preprocessor),
        ('model', LinearRegression())
    ]
)

linear_model.fit(X_train, y_train)

print("Linear Regression model trained successfully.")
```

```python
y_pred_linear = linear_model.predict(X_test)

print("First 10 predictions:")
print(y_pred_linear[:10])
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

mae_linear = mean_absolute_error(y_test, y_pred_linear)

rmse_linear = np.sqrt(
    mean_squared_error(y_test, y_pred_linear)
)

r2_linear = r2_score(y_test, y_pred_linear)

print("Linear Regression Results")
print("-------------------------")
print("MAE :", round(mae_linear, 3))
print("RMSE:", round(rmse_linear, 3))
print("R²  :", round(r2_linear, 3))
```

```python
from sklearn.tree import DecisionTreeRegressor

decision_tree_model = Pipeline(
    steps=[
        ('preprocessor', preprocessor),
        ('model', DecisionTreeRegressor(
            random_state=42,
            max_depth=4
        ))
    ]
)

decision_tree_model.fit(X_train, y_train)

print("Decision Tree model trained successfully.")
```

```python
y_pred_tree = decision_tree_model.predict(X_test)

print("First 10 predictions:")
print(y_pred_tree[:10])
```

```python
mae_tree = mean_absolute_error(y_test, y_pred_tree)

rmse_tree = np.sqrt(
    mean_squared_error(y_test, y_pred_tree)
)

r2_tree = r2_score(y_test, y_pred_tree)

print("Decision Tree Results")
print("--------------------")
print("MAE :", round(mae_tree, 3))
print("RMSE:", round(rmse_tree, 3))
print("R²  :", round(r2_tree, 3))
```

```python
from sklearn.ensemble import RandomForestRegressor

random_forest_model = Pipeline(
    steps=[
        ('preprocessor', preprocessor),
        ('model', RandomForestRegressor(
            n_estimators=200,
            random_state=42,
            max_depth=5
        ))
    ]
)

random_forest_model.fit(X_train, y_train)

print("Random Forest model trained successfully.")
```

```python
y_pred_rf = random_forest_model.predict(X_test)

print("First 10 predictions:")
print(y_pred_rf[:10])
```

```python
mae_rf = mean_absolute_error(y_test, y_pred_rf)

rmse_rf = np.sqrt(
    mean_squared_error(y_test, y_pred_rf)
)

r2_rf = r2_score(y_test, y_pred_rf)

print("Random Forest Results")
print("--------------------")
print("MAE :", round(mae_rf, 3))
print("RMSE:", round(rmse_rf, 3))
print("R²  :", round(r2_rf, 3))
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.pipeline import Pipeline
from xgboost import XGBRegressor

# XGBoost pipeline
xgb_pipeline = Pipeline(
    steps=[
        ('preprocessor', preprocessor),
        ('model', XGBRegressor(
            objective='reg:squarederror',
            random_state=42
        ))
    ]
)

# Hyperparameter combinations to test
param_grid = {
    'model__n_estimators': [100, 200, 300],
    'model__max_depth': [2, 3, 4],
    'model__learning_rate': [0.03, 0.05, 0.1]
}

# 5-fold cross-validation
grid_search = GridSearchCV(
    estimator=xgb_pipeline,
    param_grid=param_grid,
    cv=5,
    scoring='neg_mean_absolute_error',
    n_jobs=-1
)

grid_search.fit(X_train, y_train)

print("Best parameters:")
print(grid_search.best_params_)

print("\nBest cross-validation MAE:")
print(round(-grid_search.best_score_, 3))
```

```python
best_xgb_model = grid_search.best_estimator_

y_pred_tuned = best_xgb_model.predict(X_test)

mae_tuned = mean_absolute_error(y_test, y_pred_tuned)

rmse_tuned = np.sqrt(
    mean_squared_error(y_test, y_pred_tuned)
)

r2_tuned = r2_score(y_test, y_pred_tuned)

print("Tuned XGBoost Results")
print("---------------------")
print("MAE :", round(mae_tuned, 3))
print("RMSE:", round(rmse_tuned, 3))
print("R²  :", round(r2_tuned, 3))
```

```python
comparison = pd.DataFrame({
    'Model': [
        'Original XGBoost',
        'Tuned XGBoost'
    ],
    'MAE': [
        mae_xgb,
        mae_tuned
    ],
    'RMSE': [
        rmse_xgb,
        rmse_tuned
    ],
    'R2': [
        r2_xgb,
        r2_tuned
    ]
})

comparison
```

```python
# Experiment 2: Include previous academic grades G1 and G2

academic_features = [
    'studytime',
    'failures',
    'absences',
    'traveltime',
    'activities',
    'health',
    'G1',
    'G2'
]

X_academic = df[academic_features].copy()
y_academic = df['G3'].copy()

# Convert activities from yes/no to 1/0
X_academic['activities'] = X_academic['activities'].map({
    'yes': 1,
    'no': 0
})

print("Academic features:")
print(X_academic.head())

print("\nX shape:", X_academic.shape)
print("y shape:", y_academic.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train_academic, X_test_academic, y_train_academic, y_test_academic = train_test_split(
    X_academic,
    y_academic,
    test_size=0.20,
    random_state=42
)

print("Training data:")
print("X_train:", X_train_academic.shape)
print("y_train:", y_train_academic.shape)

print("\nTesting data:")
print("X_test:", X_test_academic.shape)
print("y_test:", y_test_academic.shape)
```

```python
from xgboost import XGBRegressor

academic_xgb_model = XGBRegressor(
    n_estimators=200,
    max_depth=3,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    random_state=42
)

academic_xgb_model.fit(
    X_train_academic,
    y_train_academic
)

print("Academic XGBoost model trained successfully.")
```

```python
y_pred_academic = academic_xgb_model.predict(
    X_test_academic
)

print("First 10 predictions:")
print(y_pred_academic[:10])
```

```python
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

import numpy as np

mae_academic = mean_absolute_error(
    y_test_academic,
    y_pred_academic
)

rmse_academic = np.sqrt(
    mean_squared_error(
        y_test_academic,
        y_pred_academic
    )
)

r2_academic = r2_score(
    y_test_academic,
    y_pred_academic
)

print("XGBoost with G1 + G2 Results")
print("-----------------------------")
print("MAE :", round(mae_academic, 3))
print("RMSE:", round(rmse_academic, 3))
print("R²  :", round(r2_academic, 3))
```

```python
from sklearn.model_selection import RandomizedSearchCV
from xgboost import XGBRegressor
import numpy as np

# Base XGBoost model
xgb_base = XGBRegressor(
    objective='reg:squarederror',
    random_state=42
)

# Hyperparameter search space
param_grid = {
    'n_estimators': [100, 150, 200, 250, 300, 400],
    'max_depth': [2, 3, 4, 5],
    'learning_rate': [0.01, 0.03, 0.05, 0.07, 0.1],
    'subsample': [0.7, 0.8, 0.9, 1.0],
    'colsample_bytree': [0.7, 0.8, 0.9, 1.0],
    'min_child_weight': [1, 2, 3, 5],
    'gamma': [0, 0.1, 0.2, 0.3],
    'reg_alpha': [0, 0.01, 0.1],
    'reg_lambda': [1, 1.5, 2, 3]
}

# Randomized search
random_search = RandomizedSearchCV(
    estimator=xgb_base,
    param_distributions=param_grid,
    n_iter=30,
    scoring='neg_mean_absolute_error',
    cv=5,
    random_state=42,
    n_jobs=-1,
    verbose=1
)

# IMPORTANT:
# Only training data is used for tuning
random_search.fit(
    X_train_academic,
    y_train_academic
)

print("Hyperparameter tuning completed successfully.")
```

```python
print("Best Parameters:")
print(random_search.best_params_)

print("\nBest Cross-Validation MAE:")
print(round(-random_search.best_score_, 3))
```

```python
tuned_xgb_model = random_search.best_estimator_

print("Tuned XGBoost model obtained successfully.")
```

```python
y_pred_tuned = tuned_xgb_model.predict(
    X_test_academic
)

print("First 10 tuned predictions:")
print(y_pred_tuned[:10])
```

```python
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

mae_tuned = mean_absolute_error(
    y_test_academic,
    y_pred_tuned
)

rmse_tuned = np.sqrt(
    mean_squared_error(
        y_test_academic,
        y_pred_tuned
    )
)

r2_tuned = r2_score(
    y_test_academic,
    y_pred_tuned
)

print("Tuned XGBoost Results")
print("---------------------")
print("MAE :", round(mae_tuned, 3))
print("RMSE:", round(rmse_tuned, 3))
print("R²  :", round(r2_tuned, 3))
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 6))

plt.scatter(
    y_test_academic,
    y_pred_tuned,
    alpha=0.7
)

# Perfect prediction line
plt.plot(
    [0, 20],
    [0, 20],
    linestyle='--'
)

plt.xlabel("Actual Final Grade (G3)")
plt.ylabel("Predicted Final Grade (G3)")
plt.title("Actual vs Predicted Final Grade - Tuned XGBoost")

plt.xlim(0, 20)
plt.ylim(0, 20)

plt.grid(alpha=0.2)

plt.show()
```

```python
# Convert predicted G3 to percentage
predicted_percentage = (y_pred_tuned / 20) * 100

# PASS / FAIL classification
predicted_status = np.where(
    y_pred_tuned >= 10,
    "PASS",
    "FAIL"
)

# Actual PASS / FAIL
actual_status = np.where(
    y_test_academic >= 10,
    "PASS",
    "FAIL"
)

print("First 10 predictions:")
for i in range(10):
    print(
        f"Predicted G3: {y_pred_tuned[i]:.2f} | "
        f"Percentage: {predicted_percentage[i]:.2f}% | "
        f"Status: {predicted_status[i]} | "
        f"Actual: {actual_status[i]}"
    )
```

```python
from sklearn.metrics import accuracy_score

pass_fail_accuracy = accuracy_score(
    actual_status,
    predicted_status
)

print(
    "PASS/FAIL Accuracy:",
    round(pass_fail_accuracy * 100, 2),
    "%"
)
```
